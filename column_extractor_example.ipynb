```python
import matplotlib.pyplot as plt
import matplotlib.dates as plt_dates
import numpy as np
import xarray as xr
```

```python
xr.set_options(display_max_rows=64)
```

```python
data = xr.open_dataset("column_extractor_example.nc")
#print(data)
```

```python
salinity = data["salt"]
#print(salinity)
```

```python
depth_from_msl = data["z"]
#print(depth_from_msl.values)
```

```python
time = data["time"]
time_labels = time.dt.strftime("%d %b %Y, %H:%M")
#print(time.values)
```

```python
# plot salinity vs depth at a given time
record = 11  # which time record to plot

fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(1, 1, 1)

ax.plot(salinity.isel(nrecords=record), depth_from_msl)

ax.grid(linestyle="--", alpha=0.25)
ax.set_title("Salinity vs Depth at " + str(time_labels.values[record]))
ax.set_xlabel("Salinity [psu]")
ax.set_ylabel("Depth from MSL [m]")
```

```python
# plot salinity over time at a given depth
level = 28  # which depth level to plot

fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(1, 1, 1)

ax.plot(time_labels, salinity.isel(k=level))

ax.grid(linestyle="--", alpha=0.25)
ax.set_title("Evolution of Salinity at z=" + str(depth_from_msl.values[level]))
ax.set_ylabel("Salinity [psu]")
ax.tick_params(axis="x", labelrotation=75)
```
