```python
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr
```

```python
xr.set_options(display_max_rows=64)
```

```python
PT_ACTIVE = 0x001  # particle is active
```

```python
data = xr.open_dataset("particle_tracking_example.nc")
#print(data)
```

```python
fig, ax = plt.subplots(3, 1, figsize=(8, 24), layout="constrained", subplot_kw={"projection": "3d", "proj_type": "ortho"})

ax[0].set_xlabel("Lon [deg]", fontsize=10)
ax[0].set_ylabel("Lat [deg]", fontsize=10)
ax[0].set_zlabel("Depth [m]", fontsize=10)

ax[1].set_xlabel("Lon [deg]", fontsize=10)
ax[1].set_ylabel("Lat [deg]", fontsize=10)
ax[1].set_zlabel("Depth [m]", fontsize=10)

ax[2].set_xlabel("Lon [deg]", fontsize=10)
ax[2].set_ylabel("Lat [deg]", fontsize=10)
ax[2].set_zlabel("Depth [m]", fontsize=10)

ax[0].view_init(elev=90, azim=-90, roll=0)  # xy plane
ax[1].view_init(elev=0., azim=-90, roll=0)  # xz plane

num_tracks = int(data.sizes["n"])
for track in range(num_tracks):
    # extract [lon, lat, z] particle positions for a given track
    xlon = data["x"][:, track]
    ylat = data["y"][:, track]
    zpos = data["z"][:, track]

    # keep the parts of the arrays in which particles are active
    flag = data["flag"][:, track]
    okay = flag == PT_ACTIVE
    
    ax[0].plot(xlon[okay], ylat[okay], zpos[okay])
    ax[1].plot(xlon[okay], ylat[okay], zpos[okay])
    ax[2].plot(xlon[okay], ylat[okay], zpos[okay])
```
