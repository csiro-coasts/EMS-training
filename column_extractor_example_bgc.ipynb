```python
import matplotlib.pyplot as plt
import matplotlib.dates as plt_dates
import numpy as np
import xarray as xr
```

```python
xr.set_options(display_max_rows=64)
```

```python
data_sb1 = xr.open_dataset("sb1.nc")
data_sb2 = xr.open_dataset("sb2.nc")
data_sb3 = xr.open_dataset("sb3.nc")
data_sb4 = xr.open_dataset("sb4.nc")
#print(data_sb1)
```

```python
chl_a_sb1 = data_sb1["Chl_a_sum"].mean(dim="nrecords")
chl_a_sb2 = data_sb2["Chl_a_sum"].mean(dim="nrecords")
chl_a_sb3 = data_sb3["Chl_a_sum"].mean(dim="nrecords")
chl_a_sb4 = data_sb4["Chl_a_sum"].mean(dim="nrecords")
```

```python
DIN_sb1 = data_sb1["NO3"].mean(dim="nrecords") + \
          data_sb1["NH4"].mean(dim="nrecords")
DIN_sb2 = data_sb2["NO3"].mean(dim="nrecords") + \
          data_sb2["NH4"].mean(dim="nrecords")
DIN_sb3 = data_sb3["NO3"].mean(dim="nrecords") + \
          data_sb3["NH4"].mean(dim="nrecords")
DIN_sb4 = data_sb4["NO3"].mean(dim="nrecords") + \
          data_sb4["NH4"].mean(dim="nrecords")
```

```python
temperature_sb1 = data_sb1["temp"].mean(dim="nrecords")
temperature_sb2 = data_sb2["temp"].mean(dim="nrecords")
temperature_sb3 = data_sb3["temp"].mean(dim="nrecords")
temperature_sb4 = data_sb4["temp"].mean(dim="nrecords")
```

```python
depth_from_msl_sb1 = data_sb1["z"]
depth_from_msl_sb2 = data_sb2["z"]
depth_from_msl_sb3 = data_sb3["z"]
depth_from_msl_sb4 = data_sb4["z"]
```

```python
fig = plt.figure(figsize=(14, 5))

ax = fig.add_subplot(1, 3, 1)
ax.plot(chl_a_sb1, depth_from_msl_sb1, color="red")
ax.plot(chl_a_sb2, depth_from_msl_sb2, color="darkorange")
ax.plot(chl_a_sb3, depth_from_msl_sb3, color="yellowgreen")
ax.plot(chl_a_sb4, depth_from_msl_sb4, color="gold")

ax.grid(linestyle="--", alpha=0.25)
ax.set_title("Chl_a_sum vs Depth")
ax.set_xlabel("mg Chl m^-3")
ax.set_ylabel("Depth from MSL [m]")

ax = fig.add_subplot(1, 3, 2)
ax.plot(DIN_sb1, depth_from_msl_sb1, color="red")
ax.plot(DIN_sb2, depth_from_msl_sb2, color="darkorange")
ax.plot(DIN_sb3, depth_from_msl_sb3, color="yellowgreen")
ax.plot(DIN_sb4, depth_from_msl_sb4, color="gold")

ax.grid(linestyle="--", alpha=0.25)
ax.set_title("(NO3+NH4) vs Depth")
ax.set_xlabel("mg N m^-3")

ax = fig.add_subplot(1, 3, 3)
ax.plot(temperature_sb1, depth_from_msl_sb1, color="red")
ax.plot(temperature_sb2, depth_from_msl_sb2, color="darkorange")
ax.plot(temperature_sb3, depth_from_msl_sb3, color="yellowgreen")
ax.plot(temperature_sb4, depth_from_msl_sb4, color="gold")

ax.grid(linestyle="--", alpha=0.25)
ax.set_title("Temperature vs Depth")
ax.set_xlabel("deg C")

ax.legend(("SB1", "SB2", "SB3", "SB4"));
```
